# Mutual information as departure from independence

Executable examples corresponding to the paper.

```python
import math
from markov_entropy import FiniteSpace, Distribution, mutual_information
from markov_entropy.divergences import KL
X = FiniteSpace([0, 1])
p_xy = Distribution(X.tensor(X), [0.5, 0, 0, 0.5])
mutual_information(p_xy, KL()), math.log(2)
```
