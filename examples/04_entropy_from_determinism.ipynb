# Entropy as departure from determinism

Executable examples corresponding to the paper.

```python
from markov_entropy import Distribution, FiniteSpace, entropy, gini_simpson
from markov_entropy.divergences import KL, Renyi, TotalVariation

X = FiniteSpace([0, 1])
p = Distribution(X, [0.25, 0.75])
entropy(p, KL()), entropy(p, Renyi(0.5)), entropy(p, TotalVariation()), gini_simpson(p)
```
