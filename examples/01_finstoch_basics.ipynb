# FinStoch basics

Executable examples corresponding to the paper.

```python
from markov_entropy import Channel, Distribution, FiniteSpace, joint

X = FiniteSpace([0, 1])
p = Distribution(X, [0.25, 0.75])
f = Channel(X, X, [[0.9, 0.2], [0.1, 0.8]])
xy = joint(p, f)
xy.probabilities, xy.marginal(0).probabilities, xy.marginal(1).probabilities
```
